## Extracting latest news of bajar ko chirfaar 

```python
https://bajarkochirfar.com/latest/


# /html/body/div[1]/header/div[2]/button[1]  : click 



# /html/body/div[1]/section/div/div[2]/div[4]/article/a[2]



# /html/body/div[1]/div[1]/div/div[1]/main/div/div[1]/article/div 

# title
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
from datetime import datetime
import pandas as pd
import os
import json

# 1. Initialize the Chrome WebDriver instance
driver = webdriver.Chrome()

driver.get("https://bajarkochirfar.com/latest/")
print(f"Page Title: {driver.title}")


driver.execute_script("window.scrollBy(0, 1000);")

time.sleep(5)

driver.find_element(By.XPATH,"/html/body/div[1]/header/div[2]/button[1]").click()

time.sleep(5)
url_links=[]

for i in range(1,6):
    try:
        url=driver.find_element(By.XPATH,f"/html/body/div[1]/section/div/div[2]/div[{i}]/article/a[2]").get_attribute("href")
        url_links.append(url)
        # print(url)
    except:
        print("pass")

# /html/body/div[1]/div[1]/div/div[1]/main/div/div[1]/article/div
news={}
for url in url_links:
    driver.get(url)
    time.sleep(5)
    news[driver.title]=driver.find_element(By.XPATH,"/html/body/div[1]/div[1]/div/div[1]/main/div/div[1]/article/div").text
    
now = datetime.now()

# safe filename (remove spaces + colons)
news_time = now.strftime("%Y-%m-%d_%H-%M-%S")

# dataframe
df = pd.DataFrame(list(news.items()), columns=["title", "news"])

# folder path
folder = r"C:\Users\HP\Desktop\data_extractions\New folder\latest_news"
os.makedirs(folder, exist_ok=True)

csv_path = os.path.join(folder, f"{news_time}.csv")

df.to_csv(
    csv_path,
    index=False,
    encoding="utf-8-sig"
)


json_path = os.path.join(folder, f"{news_time}.json")

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(
        df.to_dict(orient="records"),
        f,
        ensure_ascii=False,
        indent=4
    )

driver.quit()
```
