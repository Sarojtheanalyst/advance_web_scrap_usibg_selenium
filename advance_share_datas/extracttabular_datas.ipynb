# Code to implement or extract the tabuklar datas 

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC



import pandas as pd
import numpy as np
import time
import os
```

```python
wait = WebDriverWait(driver, 20)

wait.until(
    EC.presence_of_element_located((By.TAG_NAME, "table"))
)

time.sleep(3)

# =========================================================
# FIND ALL TABLES
# =========================================================

tables = driver.find_elements(By.TAG_NAME, "table")

print(f"\nTOTAL TABLES FOUND: {len(tables)}")

# =========================================================
# STORE ALL DATAFRAMES
# =========================================================

all_tables = {}

# =========================================================
# LOOP THROUGH TABLES
# =========================================================

for table_index, table in enumerate(tables):

    print(f"\n{'='*60}")
    print(f"EXTRACTING TABLE {table_index+1}")
    print(f"{'='*60}")

    try:

        # =================================================
        # FIND ROWS
        # =================================================

        rows = table.find_elements(By.TAG_NAME, "tr")

        print(f"TOTAL ROWS: {len(rows)}")

        extracted_rows = []

        # =================================================
        # LOOP THROUGH ROWS
        # =================================================

        for row_index, row in enumerate(rows):

            row_data = []

            # =============================================
            # EXTRACT BOTH TH AND TD
            # =============================================

            cells = row.find_elements(
                By.XPATH,
                ".//th | .//td"
            )

            # =============================================
            # SKIP EMPTY ROWS
            # =============================================

            if len(cells) == 0:
                continue

            # =============================================
            # EXTRACT CELL DATA
            # =============================================

            for cell_index, cell in enumerate(cells):

                try:

                    text = cell.text.strip()

                    # =====================================
                    # HANDLE EMPTY CELLS
                    # =====================================

                    if text == "":
                        text = np.nan

                    row_data.append(text)

                except Exception as cell_error:

                    print(
                        f"Cell Error: {cell_error}"
                    )

                    row_data.append(np.nan)

            # =============================================
            # STORE ROW
            # =============================================

            if len(row_data) > 0:

                extracted_rows.append(row_data)

                print(
                    f"ROW {row_index+1}: {row_data}"
                )

        # =================================================
        # SKIP EMPTY TABLE
        # =================================================

        if len(extracted_rows) == 0:
            continue

        # =================================================
        # DETECT MAX COLUMNS
        # =================================================

        max_cols = max(
            len(row)
            for row in extracted_rows
        )

        # =================================================
        # NORMALIZE ROW LENGTHS
        # =================================================

        normalized_rows = []

        for row in extracted_rows:

            if len(row) < max_cols:

                row.extend(
                    [np.nan] * (max_cols - len(row))
                )

            normalized_rows.append(row)

        # =================================================
        # DETECT HEADER
        # =================================================

        headers = normalized_rows[0]

        data = normalized_rows[1:]

        # =================================================
        # HANDLE DUPLICATE COLUMNS
        # =================================================

        clean_headers = []

        header_count = {}
        for col in headers:
            if pd.isna(col):
                col = "unknown"
            col = str(col).strip()
            if col in header_count:
                header_count[col] += 1
                col = f"{col}_{header_count[col]}"

            else:

                header_count[col] = 1

            clean_headers.append(col)

        df = pd.DataFrame(
            data,
            columns=clean_headers
        )

        df.columns = (
            df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
            .str.replace("-", "_")
        )

        df.drop_duplicates(inplace=True)
        df.reset_index(drop=True, inplace=True)s
        df.dropna(
            how="all",
            inplace=True
        )

        df = df.apply(
            lambda col: col.astype(str).str.strip()
        )

        table_name = f"table_{table_index+1}"

        all_tables[table_name] = df

        print(f"\nDATAFRAME: {table_name}")

        print(df.head())

        print("\nSHAPE:", df.shape)

    except Exception as table_error:

        print(
            f"\nERROR IN TABLE {table_index+1}"
        )

        print(table_error)

excel_file = "all_extracted_tables.xlsx"

with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    for table_name, df in all_tables.items():

        df.to_excel(
            writer,
            sheet_name=table_name[:31],
            index=False
        )

print(f"\nALL TABLES SAVED TO: {excel_file}")

# =========================================================
# SAVE CSV FILES
# =========================================================

os.makedirs(
    "tables_csv",
    exist_ok=True
)

for table_name, df in all_tables.items():

    csv_path = f"tables_csv/{table_name}.csv"

    df.to_csv(
        csv_path,
        index=False
    )

print("\nALL CSV FILES SAVED")
```
