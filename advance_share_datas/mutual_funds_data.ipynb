```python
# closed ended 
all_rows=[]
for i in range(1,45):
    rows=[]
    for j in range(1,11):
        try:
            ele=driver.find_element(By.XPATH,f"//div//div//div//div//div[2]//div//table//tbody//tr[{i}]//td[{j}]")
            # print(ele.text)
            rows.append(ele.text)
        except:
            rows.append(None)
    all_rows.append(rows)
print("Extracted all the data sucessfully ")
```
